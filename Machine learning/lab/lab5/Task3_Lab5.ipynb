## K-Nearest Neighbors (KNN)

Assumes proximity equals similarity: data points close to each other usually belong to the same category.

Requires feature scaling, since calculated distances directly dictate the outcome.

Treats all features as having equal influence unless explicitly weighted.

## Decision Tree

Splits data one variable at a time using orthogonal (axis-aligned) thresholds.

Uses a greedy approach, selecting the best split at each step without planning ahead.

Assumes training samples are independent and adequately reflect real-world distributions.

Non-parametric and scale-invariant, meaning it requires neither normalized features nor linear relationships.

Prone to severe overfitting if depth constraints or pruning techniques are omitted.

## Random Forest

Inherits the baseline properties and assumptions of standard decision trees.

Trains individual trees on random subsets of data and features to decorrelate them and reduce variance.

Assumes independent, representative training observations.

Remains robust against unscaled features and non-normal data distributions.
